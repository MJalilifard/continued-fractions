```julia
a = [10,20,30]
```

```julia
function allPermute(a)
    function permuteRight(a)
        a2 = copy(a)
        b = zeros(Int64,length(a))
        for i in 1:(length(a)-1)
            b[i+1] = a2[i]
        end
        b[1] = a2[end]
        return b
    end
        global allPerms = []
        for i in 1:(length(a))
            push!(allPerms,permuteRight(a))
            a = permuteRight(a)
        end
        return allPerms
end
```

```julia
allPermute(a)
```

```julia
allPermute([1,5,9,7,5,9])
```
